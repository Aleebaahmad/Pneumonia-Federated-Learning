```python
dataset_path = r"C:\Pneumonia\X-ray"

train_hashes = {}
test_hashes = {}

# Hash training images
for class_name in ["NORMAL", "PNEUMONIA"]:
    class_path = os.path.join(dataset_path, "train", class_name)

    for filename in os.listdir(class_path):
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):

            file_path = os.path.join(class_path, filename)

            with open(file_path, "rb") as f:
                file_hash = hashlib.md5(f.read()).hexdigest()

            train_hashes[file_hash] = file_path


# Hash testing images
for class_name in ["NORMAL", "PNEUMONIA"]:
    class_path = os.path.join(dataset_path, "test", class_name)

    for filename in os.listdir(class_path):
        if filename.lower().endswith((".jpg", ".jpeg", ".png")):

            file_path = os.path.join(class_path, filename)

            with open(file_path, "rb") as f:
                file_hash = hashlib.md5(f.read()).hexdigest()

            test_hashes[file_hash] = file_path


# Find overlap
overlap = set(train_hashes.keys()) & set(test_hashes.keys())

print("Training images:", len(train_hashes))
print("Testing images:", len(test_hashes))
print("Exact duplicates between training and testing:", len(overlap))

if overlap:
    print("\nDuplicated images:")
    for h in overlap:
        print("TRAIN:", train_hashes[h])
        print("TEST :", test_hashes[h])
        print()
```

```python
import os
import hashlib
from collections import defaultdict

def get_file_hash(filepath):
    with open(filepath, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

# Get paths
train_paths = train_dataset.file_paths
test_paths = test_dataset.file_paths

# Hash all images
train_hashes = defaultdict(list)
test_hashes = defaultdict(list)

for path in train_paths:
    train_hashes[get_file_hash(path)].append(path)

for path in test_paths:
    test_hashes[get_file_hash(path)].append(path)

# Exact duplicates within training
train_duplicate_groups = {
    h: paths for h, paths in train_hashes.items()
    if len(paths) > 1
}

# Exact duplicates within testing
test_duplicate_groups = {
    h: paths for h, paths in test_hashes.items()
    if len(paths) > 1
}

# Exact duplicates across train and test
cross_split_duplicates = []

for h in train_hashes:
    if h in test_hashes:
        cross_split_duplicates.append(
            (train_hashes[h], test_hashes[h])
        )

print("===== EXACT DUPLICATE ANALYSIS =====")

print("Training images:", len(train_paths))
print("Testing images:", len(test_paths))

print("\nDuplicate groups inside TRAIN:", len(train_duplicate_groups))
print("Duplicate groups inside TEST:", len(test_duplicate_groups))

print("Exact duplicates across TRAIN ↔ TEST:",
      len(cross_split_duplicates))

if cross_split_duplicates:
    print("\nExamples of train/test duplicates:")

    for train_files, test_files in cross_split_duplicates[:10]:
        print("\nTRAIN:")
        for f in train_files:
            print(" ", f)

        print("TEST:")
        for f in test_files:
            print(" ", f)
```

Near Duplication analysis

```python
import cv2
import numpy as np
from collections import defaultdict

def perceptual_hash(image_path):
    """
    Generate a simple perceptual hash using DCT.
    Similar-looking images should have similar hashes.
    """
    
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if img is None:
        return None

    # Resize to 32x32
    img = cv2.resize(img, (32, 32))

    # DCT
    dct = cv2.dct(np.float32(img))

    # Take low-frequency components
    dct_low = dct[:8, :8]

    # Remove DC component
    dct_values = dct_low.flatten()[1:]

    # Median threshold
    median = np.median(dct_values)

    hash_bits = dct_values > median

    return hash_bits


def hamming_distance(hash1, hash2):
    return np.sum(hash1 != hash2)


# Generate hashes
train_phashes = {}
test_phashes = {}

print("Generating TRAIN perceptual hashes...")

for path in train_paths:
    h = perceptual_hash(path)
    if h is not None:
        train_phashes[path] = h

print("Generating TEST perceptual hashes...")

for path in test_paths:
    h = perceptual_hash(path)
    if h is not None:
        test_phashes[path] = h


# Find near-duplicates across TRAIN and TEST
near_duplicates_cross_split = []

for test_path, test_hash in test_phashes.items():

    for train_path, train_hash in train_phashes.items():

        distance = hamming_distance(test_hash, train_hash)

        if distance <= 5:
            near_duplicates_cross_split.append(
                (distance, train_path, test_path)
            )


# Sort by similarity
near_duplicates_cross_split.sort(key=lambda x: x[0])


print("\n===== NEAR-DUPLICATE ANALYSIS =====")

print(
    "Potential near-duplicates across TRAIN ↔ TEST:",
    len(near_duplicates_cross_split)
)

print("\nMost similar pairs:")

for distance, train_file, test_file in near_duplicates_cross_split[:20]:

    print("\nHamming distance:", distance)
    print("TRAIN:", train_file)
    print("TEST :", test_file)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# ============================================
# THRESHOLD ANALYSIS
# ============================================

thresholds = np.arange(0.10, 0.91, 0.05)

results = []

for threshold in thresholds:

    # Convert probabilities into predictions
    y_pred_threshold = (y_prob >= threshold).astype(int)

    # Confusion matrix
    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_pred_threshold,
        labels=[0, 1]
    ).ravel()

    # Metrics
    accuracy = accuracy_score(y_true, y_pred_threshold)

    precision = precision_score(
        y_true,
        y_pred_threshold,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred_threshold,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred_threshold,
        zero_division=0
    )

    # Specificity = TN / (TN + FP)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })


# Convert results to DataFrame
threshold_results = pd.DataFrame(results)

print("==============================================")
print("          THRESHOLD ANALYSIS")
print("==============================================")

print(
    threshold_results.round(4).to_string(index=False)
)
```

Building Tensorflow dataset

```python
import tensorflow as tf

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

# ============================================
# CREATE DATASET FROM PATHS
# ============================================

def load_image(path, label):

    image = tf.io.read_file(path)

    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    image = tf.image.resize(
        image,
        IMG_SIZE
    )

    image = tf.cast(
        image,
        tf.float32
    )

    return image, tf.cast(label, tf.float32)


# Training dataset
train_ds = tf.data.Dataset.from_tensor_slices(
    (train_paths_split, train_labels_split)
)

train_ds = train_ds.shuffle(
    buffer_size=len(train_paths_split),
    seed=42
)

train_ds = train_ds.map(
    load_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

train_ds = train_ds.batch(BATCH_SIZE)
train_ds = train_ds.prefetch(tf.data.AUTOTUNE)


# Validation dataset
val_ds = tf.data.Dataset.from_tensor_slices(
    (val_paths_split, val_labels_split)
)

val_ds = val_ds.map(
    load_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

val_ds = val_ds.batch(BATCH_SIZE)
val_ds = val_ds.prefetch(tf.data.AUTOTUNE)


print("Training and validation datasets created.")
```

performing threshold analysis

```python
# ============================================
# VALIDATION THRESHOLD ANALYSIS
# ============================================

thresholds = np.arange(0.10, 0.91, 0.05)

results = []

for threshold in thresholds:

    predictions = (
        val_prob >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        val_true,
        predictions,
        labels=[0, 1]
    ).ravel()

    accuracy = accuracy_score(
        val_true,
        predictions
    )

    precision = precision_score(
        val_true,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        val_true,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        val_true,
        predictions,
        zero_division=0
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0
    )

    results.append({
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })


threshold_results = pd.DataFrame(results)

print("\n==============================================")
print("       VALIDATION THRESHOLD ANALYSIS")
print("==============================================")

print(
    threshold_results.round(4).to_string(index=False)
)
```

Creating clean set of dataset

```python
import tensorflow as tf
from pathlib import Path

# ============================================
# CLEAN TEST DATASET
# ============================================

TEST_DIR = Path(r"C:\Pneumonia\X-ray\test_clean")

IMG_SIZE = (224, 224)
BATCH_SIZE = 32

test_paths = []
test_labels = []

class_mapping = {
    "NORMAL": 0,
    "PNEUMONIA": 1
}

for class_name, label in class_mapping.items():

    class_dir = TEST_DIR / class_name

    for file in class_dir.rglob("*"):

        if file.is_file() and file.suffix.lower() in [
            ".jpg",
            ".jpeg",
            ".png"
        ]:

            test_paths.append(str(file))
            test_labels.append(label)


print("======================================")
print("       CLEAN TEST DATASET")
print("======================================")

print("NORMAL:",
      sum(np.array(test_labels) == 0))

print("PNEUMONIA:",
      sum(np.array(test_labels) == 1))

print("TOTAL:",
      len(test_labels))


# ============================================
# LOAD IMAGES
# ============================================

def load_test_image(path, label):

    image = tf.io.read_file(path)

    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    image = tf.image.resize(
        image,
        IMG_SIZE
    )

    image = tf.cast(
        image,
        tf.float32
    )

    return image, tf.cast(label, tf.float32)


test_ds = tf.data.Dataset.from_tensor_slices(
    (
        test_paths,
        test_labels
    )
)

test_ds = test_ds.map(
    load_test_image,
    num_parallel_calls=tf.data.AUTOTUNE
)

test_ds = test_ds.batch(BATCH_SIZE)

test_ds = test_ds.prefetch(
    tf.data.AUTOTUNE
)

print("\nClean test dataset created successfully.")
```

Generating complete error analysis table

```python
# ============================================
# FINAL TEST ERROR ANALYSIS DATAFRAME
# ============================================

import pandas as pd
import numpy as np
from pathlib import Path

# Create results dataframe
error_df = pd.DataFrame({
    "filepath": test_paths,
    "true_label": test_true,
    "probability": test_prob,
    "predicted_label": test_pred
})

# Convert numeric labels to names
label_names = {
    0: "NORMAL",
    1: "PNEUMONIA"
}

error_df["true_class"] = error_df["true_label"].map(label_names)
error_df["predicted_class"] = error_df["predicted_label"].map(label_names)

# Mark errors
error_df["correct"] = (
    error_df["true_label"] ==
    error_df["predicted_label"]
)

# Error type
def get_error_type(row):

    if row["true_label"] == 0 and row["predicted_label"] == 1:
        return "FALSE_POSITIVE"

    elif row["true_label"] == 1 and row["predicted_label"] == 0:
        return "FALSE_NEGATIVE"

    elif row["true_label"] == 0 and row["predicted_label"] == 0:
        return "TRUE_NEGATIVE"

    elif row["true_label"] == 1 and row["predicted_label"] == 1:
        return "TRUE_POSITIVE"

error_df["error_type"] = error_df.apply(
    get_error_type,
    axis=1
)

print("==============================================")
print("             ERROR ANALYSIS")
print("==============================================")

print("\nTotal images:", len(error_df))

print("\nCorrect predictions:",
      error_df["correct"].sum())

print("Incorrect predictions:",
      (~error_df["correct"]).sum())

print("\nError breakdown:")
print(
    error_df["error_type"]
    .value_counts()
)

print("\nExpected:")
print("TN:", 107)
print("FP:", 119)
print("FN:", 7)
print("TP:", 379)
```

```python
# ============================================
# WEIGHTED MODEL THRESHOLD ANALYSIS
# ============================================

threshold_results_weighted = []

thresholds = np.arange(0.10, 0.91, 0.01)

for threshold in thresholds:

    predictions = (
        val_prob_weighted >= threshold
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        val_true_weighted,
        predictions,
        labels=[0, 1]
    ).ravel()

    accuracy = accuracy_score(
        val_true_weighted,
        predictions
    )

    precision = precision_score(
        val_true_weighted,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        val_true_weighted,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        val_true_weighted,
        predictions,
        zero_division=0
    )

    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0

    threshold_results_weighted.append({
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Specificity": specificity,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })


threshold_df_weighted = pd.DataFrame(
    threshold_results_weighted
)

print("==============================================")
print("     WEIGHTED MODEL THRESHOLD ANALYSIS")
print("==============================================")

print(
    threshold_df_weighted.round(4).to_string(index=False)
)
```

Creating Grad cam function

```python
# ============================================
# GRAD-CAM FUNCTION
# ============================================

def make_gradcam_heatmap(
    image,
    model,
    last_conv_layer_name="efficientnetb0"
):

    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[
            model.get_layer(last_conv_layer_name).output,
            model.output
        ]
    )

    image = tf.cast(image, tf.float32)

    if len(image.shape) == 3:
        image = tf.expand_dims(image, axis=0)

    with tf.GradientTape() as tape:

        conv_outputs, predictions = grad_model(
            image,
            training=False
        )

        # Binary classification:
        # output = probability of PNEUMONIA
        class_score = predictions[:, 0]

    gradients = tape.gradient(
        class_score,
        conv_outputs
    )

    # Global average pooling of gradients
    pooled_gradients = tf.reduce_mean(
        gradients,
        axis=(1, 2)
    )

    conv_outputs = conv_outputs[0]
    pooled_gradients = pooled_gradients[0]

    # Weight feature maps by importance
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_gradients,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(
        heatmap,
        0
    )

    # Normalize
    max_value = tf.reduce_max(heatmap)

    heatmap = tf.where(
        max_value > 0,
        heatmap / max_value,
        heatmap
    )

    return heatmap.numpy()
```

```python

import matplotlib.pyplot as plt
import cv2

# ============================================================
# GRAD-CAM FOR YOUR EFFICIENTNETB0 MODEL
# ============================================================

def make_gradcam_heatmap(image, model, backbone_name="efficientnetb0"):

    # Find EfficientNet backbone
    backbone = model.get_layer(backbone_name)

    # Create a model that returns:
    # 1. EfficientNet feature maps
    # 2. Final model prediction
    grad_model = tf.keras.Model(
        inputs=model.inputs,
        outputs=[backbone.output, model.output]
    )

    # Ensure batch dimension
    if len(image.shape) == 3:
        image = tf.expand_dims(image, axis=0)

    image = tf.cast(image, tf.float32)

    # Forward + gradients
    with tf.GradientTape() as tape:

        conv_outputs, predictions = grad_model(
            image,
            training=False
        )

        # Probability of PNEUMONIA
        pneumonia_score = predictions[:, 0]

    # Gradient of pneumonia score
    grads = tape.gradient(
        pneumonia_score,
        conv_outputs
    )

    # Global average pooling of gradients
    pooled_grads = tf.reduce_mean(
        grads,
        axis=(1, 2)
    )

    # Remove batch dimension
    conv_outputs = conv_outputs[0]
    pooled_grads = pooled_grads[0]

    # Weighted feature maps
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(heatmap, 0)

    # Normalize
    max_value = tf.reduce_max(heatmap)

    if max_value > 0:
        heatmap /= max_value

    return heatmap.numpy()


print("Grad-CAM function created successfully.")
```

```python
# ============================================================
# GRAD-CAM FUNCTION
# ============================================================

def generate_gradcam(image, weighted_model):

    backbone = weighted_model.get_layer("efficientnetb0")
    gap = weighted_model.get_layer(
        "global_average_pooling2d_4"
    )
    dense = weighted_model.get_layer("dense_4")

    # --------------------------------------------------------
    # Make sure image has batch dimension
    # --------------------------------------------------------

    if len(image.shape) == 3:
        image = tf.expand_dims(image, axis=0)

    image = tf.cast(image, tf.float32)

    # --------------------------------------------------------
    # Forward pass through EfficientNet
    # --------------------------------------------------------

    with tf.GradientTape() as tape:

        feature_maps = backbone_grad_model(
            image,
            training=False
        )

        tape.watch(feature_maps)

        # Global average pooling
        pooled = tf.reduce_mean(
            feature_maps,
            axis=[1, 2]
        )

        # Dense layer
        predictions = dense(pooled)

        pneumonia_score = predictions[:, 0]

    # --------------------------------------------------------
    # Gradients
    # --------------------------------------------------------

    gradients = tape.gradient(
        pneumonia_score,
        feature_maps
    )

    # --------------------------------------------------------
    # Global average of gradients
    # --------------------------------------------------------

    weights = tf.reduce_mean(
        gradients,
        axis=[1, 2]
    )

    # --------------------------------------------------------
    # Weighted feature maps
    # --------------------------------------------------------

    heatmap = tf.reduce_sum(
        feature_maps * weights[:, tf.newaxis, tf.newaxis, :],
        axis=-1
    )

    # Remove batch dimension
    heatmap = heatmap[0]

    # ReLU
    heatmap = tf.maximum(
        heatmap,
        0
    )

    # Normalize
    max_value = tf.reduce_max(heatmap)

    if max_value > 0:
        heatmap = heatmap / max_value

    return heatmap.numpy()


print("Grad-CAM function created successfully.")
```

```python
# ============================================================
# GRAD-CAM VISUALIZATION OF FALSE POSITIVES
# ============================================================

import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from PIL import Image

print("==============================================")
print("       GRAD-CAM FALSE POSITIVE ANALYSIS")
print("==============================================")

# ------------------------------------------------------------
# 1. SETTINGS
# ------------------------------------------------------------

frozen_threshold = 0.40
num_images = 12

# ------------------------------------------------------------
# 2. FIND FALSE POSITIVES
# ------------------------------------------------------------

false_positive_indices = np.where(
    (test_true_weighted == 0) &
    (test_prob_weighted >= frozen_threshold)
)[0]

print("Total false positives:", len(false_positive_indices))

# Sort from most confident to least confident
sorted_fp_indices = false_positive_indices[
    np.argsort(
        test_prob_weighted[false_positive_indices]
    )[::-1]
]

top_fp_indices = sorted_fp_indices[:num_images]

print("Images selected for Grad-CAM:", len(top_fp_indices))


# ============================================================
# 3. GRAD-CAM FUNCTION
# ============================================================

def generate_gradcam(
    image,
    backbone_gradcam_model,
    classifier_model
):

    # Make sure image is float32
    image = tf.cast(image, tf.float32)

    # Add batch dimension
    if len(image.shape) == 3:
        image = tf.expand_dims(image, axis=0)

    # --------------------------------------------------------
    # Forward pass
    # --------------------------------------------------------

    with tf.GradientTape() as tape:

        # EfficientNet feature maps
        conv_outputs = backbone_gradcam_model(
            image,
            training=False
        )

        # Watch feature maps
        tape.watch(conv_outputs)

        # Global Average Pooling
        pooled = tf.reduce_mean(
            conv_outputs,
            axis=[1, 2]
        )

        # Get final Dense layer
        dense_layer = classifier_model.get_layer("dense_4")

        # Classification output
        predictions = dense_layer(pooled)

        # Pneumonia score
        pneumonia_score = predictions[:, 0]

    # --------------------------------------------------------
    # Calculate gradients
    # --------------------------------------------------------

    grads = tape.gradient(
        pneumonia_score,
        conv_outputs
    )

    if grads is None:
        raise ValueError(
            "Gradients are None. "
            "Check backbone_gradcam_model and weighted_model."
        )

    # --------------------------------------------------------
    # Average gradients spatially
    # --------------------------------------------------------

    pooled_grads = tf.reduce_mean(
        grads,
        axis=[1, 2]
    )

    # Remove batch dimension
    conv_outputs = conv_outputs[0]
    pooled_grads = pooled_grads[0]

    # --------------------------------------------------------
    # Weight feature maps using gradients
    # --------------------------------------------------------

    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(
        heatmap,
        0
    )

    # Normalize
    max_value = tf.reduce_max(heatmap)

    heatmap = tf.where(
        max_value > 0,
        heatmap / max_value,
        heatmap
    )

    return heatmap.numpy()


# ============================================================
# 4. VISUALIZE FALSE POSITIVES
# ============================================================

rows = 3
cols = 4

plt.figure(figsize=(16, 12))

for plot_idx, idx in enumerate(top_fp_indices):

    # --------------------------------------------------------
    # Get image path
    # --------------------------------------------------------

    path = test_clean_paths[idx]

    # --------------------------------------------------------
    # Load image
    # --------------------------------------------------------

    img = Image.open(path).convert("RGB")

    img = img.resize((224, 224))

    img_array = np.array(img).astype(np.float32)

    # --------------------------------------------------------
    # Generate Grad-CAM
    # --------------------------------------------------------

    heatmap = generate_gradcam(
        img_array,
        backbone_gradcam_model,
        weighted_model
    )

    # --------------------------------------------------------
    # Plot original image
    # --------------------------------------------------------

    ax = plt.subplot(
        rows,
        cols,
        plot_idx + 1
    )

    ax.imshow(img)

    # --------------------------------------------------------
    # Overlay Grad-CAM
    # --------------------------------------------------------

    ax.imshow(
        heatmap,
        cmap="jet",
        alpha=0.45,
        extent=(0, 224, 224, 0)
    )

    # --------------------------------------------------------
    # Information
    # --------------------------------------------------------

    probability = test_prob_weighted[idx]

    filename = path.split("\\")[-1]

    ax.set_title(
        f"FP #{plot_idx + 1}\n"
        f"P(Pneumonia) = {probability:.3f}",
        fontsize=10
    )

    ax.axis("off")


plt.suptitle(
    "Grad-CAM — Most Confident False Positives",
    fontsize=18
)

plt.tight_layout()

plt.show()


print("\n==============================================")
print("           GRAD-CAM COMPLETED")
print("==============================================")
print("False positives analyzed:", len(top_fp_indices))
print("Heatmap resolution: 7 × 7")
```
